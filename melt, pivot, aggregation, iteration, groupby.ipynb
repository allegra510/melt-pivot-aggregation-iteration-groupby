```python
import pandas as pd
breast_cancer = pd.read_csv('/Users/allegracavallo/Documents/DBMI/6018 Programming/breast-cancer.data', header = None)
breast_cancer.head()
```

```python
# data set up
breast_cancer.columns = [
    'class','age','menopause','tumor_size','inv_nodes','node_caps','deg_malig','breast','breast_quad','irradiat'
]
breast_cancer.head()
```

```python
# making column patient ID for each record using row index
breast_cancer['patient_id'] = breast_cancer.index
breast_cancer.head()
```

```python
# Melt - reshape age and tumor size into rows 
melt_data = pd.melt (
    breast_cancer,
    id_vars = ['patient_id'],
    value_vars = ['age','tumor_size'],
    var_name = 'characteristic',
    value_name = 'value'
)
melt_data.head()
```

```python
# display both melted rows for patient 0
melt_data[melt_data['patient_id'] == 0]
```

```python
# Pivot the melted data
pivot_data = melt_data.pivot(
    index = 'patient_id',
    columns = 'characteristic',
    values = 'value'
)
pivot_data.head()
```

```python
# Aggregation - count the malignancy grades and find the lowest (min) and highest (max) malignancy grade 
aggregation_data = breast_cancer['deg_malig'].aggregate(['count','min','max'])
aggregation_data
```

```python
# Iteration - loop through first five records 
for index, row in breast_cancer.head().iterrows():
    print(row['deg_malig'])
```

```python
# Groupby - count records by combination group of age and malignancy grade
grouby_data = breast_cancer.groupby(['age','deg_malig']).size()
grouby_data
```
